# 03. Model Training — Google Colab GPU

This notebook trains PyTorch BiLSTM and Transformer gesture sequence classifiers on Google Colab (T4 GPU runtime).

```python
# Config & Environment Setup
import torch
print(f"PyTorch Version: {torch.__version__}")
print(f"CUDA Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU Name: {torch.cuda.get_device_name(0)}")
```

```python
# Train BiLSTM Classifier
from src.training.train import train_model

print("=== Training BiLSTM Gesture Classifier ===")
bilstm_model, bilstm_history, best_bilstm_acc = train_model(
    model_type="lstm",
    splits_dir="data/splits",
    checkpoint_dir="checkpoints",
    figures_dir="reports/figures",
    batch_size=32,
    epochs=60,
    lr=1e-3,
    patience=10
)
print(f"Best BiLSTM Validation Accuracy: {best_bilstm_acc*100:.2f}%")
```

```python
# Train Transformer Encoder Classifier
print("=== Training Transformer Encoder Gesture Classifier ===")
trans_model, trans_history, best_trans_acc = train_model(
    model_type="transformer",
    splits_dir="data/splits",
    checkpoint_dir="checkpoints",
    figures_dir="reports/figures",
    batch_size=32,
    epochs=60,
    lr=1e-3,
    patience=10
)
print(f"Best Transformer Validation Accuracy: {best_trans_acc*100:.2f}%")
```
