# 01. Data Exploration — ISL-Speak

This notebook performs exploratory analysis on sample Indian Sign Language (ISL) gesture videos:
- Class distribution analysis
- Video metadata (FPS, frame count, resolution distribution)
- Sample frame visualization

```python
import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

SAMPLE_DIR = '../data/raw/sample'
print(f"Exploring dataset sample directory: {os.path.abspath(SAMPLE_DIR)}")
```

```python
# Analyze class distribution
class_counts = {}
video_metadata = []

for root, dirs, files in os.walk(SAMPLE_DIR):
    for file in files:
        if file.lower().endswith(('.mp4', '.avi', '.mov')):
            cls_name = os.path.basename(root)
            class_counts[cls_name] = class_counts.get(cls_name, 0) + 1
            
            vpath = os.path.join(root, file)
            cap = cv2.VideoCapture(vpath)
            fps = cap.get(cv2.CAP_PROP_FPS)
            fc = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
            w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
            h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
            cap.release()
            
            video_metadata.append({
                'video': file,
                'class': cls_name,
                'fps': fps,
                'frame_count': fc,
                'width': w,
                'height': h,
                'duration_sec': fc / fps if fps > 0 else 0
            })

df_meta = pd.DataFrame(video_metadata)
print(f"Total sample videos found: {len(df_meta)}")
df_meta.head(10)
```

```python
# Plot class distribution
if not df_meta.empty:
    plt.figure(figsize=(10, 4))
    df_meta['class'].value_counts().plot(kind='bar', color='darkcyan')
    plt.title('Sample Class Distribution')
    plt.xlabel('Sign Class')
    plt.ylabel('Video Count')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```
